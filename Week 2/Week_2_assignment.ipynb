```python
import pandas as pd
import numpy as np

# Create a dataset containing missing values
data = {
    'Student_ID': [101, 102, 103, 104, 105, 106],
    'Name': ['Rahul', 'Amit', 'Priya', 'Neha', 'Rohit', 'Anjali'],
    'Age': [20, 21, np.nan, 22, 20, np.nan],
    'Marks': [85, np.nan, 92, 88, np.nan, 95],
    'Attendance': [90, 85, 95, np.nan, 80, 98]
}

df = pd.DataFrame(data)

# 1. Display the original dataset
print("Original Dataset:")
print(df)

# 2. Identify missing values using isnull()
print("\nMissing Values:")
print(df.isnull())

# 3. Count missing values in each column
print("\nMissing Values Count:")
print(df.isnull().sum())

# 4. Remove rows containing missing values
df_removed = df.dropna()

print("\nDataset After Removing Missing Values:")
print(df_removed)

# 5. Fill missing values using appropriate methods
df_filled = df.copy()

# Fill missing Age values with the mean age
df_filled['Age'] = df_filled['Age'].fillna(df_filled['Age'].mean())

# Fill missing Marks values with the mean marks
df_filled['Marks'] = df_filled['Marks'].fillna(df_filled['Marks'].mean())

# Fill missing Attendance values with the mean attendance
df_filled['Attendance'] = df_filled['Attendance'].fillna(
    df_filled['Attendance'].mean()
)

# 6. Display the dataset after filling missing values
print("\nDataset After Filling Missing Values:")
print(df_filled)

# 7. Verify that no missing values remain
print("\nMissing Values After Filling:")
print(df_filled.isnull().sum())
```

```python
import pandas as pd

# Create the first DataFrame: Student Information
df1 = pd.DataFrame({
    'Student_ID': [101, 102, 103, 104, 105],
    'Name': ['Rahul', 'Amit', 'Priya', 'Neha', 'Rohit'],
    'Department': ['CS', 'IT', 'CS', 'IT', 'CS']
})

# Create the second DataFrame: Student Marks
df2 = pd.DataFrame({
    'Student_ID': [101, 102, 103, 104, 105],
    'Marks': [85, 78, 92, 88, 75],
    'Attendance': [90, 85, 95, 92, 80]
})

# Display original DataFrames
print("First DataFrame:")
print(df1)

print("\nSecond DataFrame:")
print(df2)


# 1. MERGE TWO DATAFRAMES
merged_df = pd.merge(df1, df2, on='Student_ID')

print("\n1. Merged DataFrame:")
print(merged_df)


# 2. CONCATENATE TWO DATAFRAMES
df3 = pd.DataFrame({
    'Student_ID': [106, 107],
    'Name': ['Anjali', 'Vikas'],
    'Department': ['IT', 'CS']
})

concatenated_df = pd.concat([df1, df3], ignore_index=True)

print("\n2. Concatenated DataFrame:")
print(concatenated_df)


# 3. GROUPBY DEPARTMENT
grouped_df = merged_df.groupby('Department')

print("\n3. GroupBy Department - Total Marks:")
print(grouped_df['Marks'].sum())


# 4. AGGREGATE FUNCTIONS
print("\n4. Aggregate Statistics by Department:")
print(
    merged_df.groupby('Department')['Marks'].agg(
        ['sum', 'mean', 'count', 'min', 'max']
    )
)


# 5. PIVOT TABLE
pivot_df = pd.pivot_table(
    merged_df,
    values='Marks',
    index='Department',
    columns='Name',
    aggfunc='mean',
    fill_value=0
)

print("\n5. Pivot Table:")
print(pivot_df)


# 6. PIVOT TABLE WITH MULTIPLE AGGREGATIONS
pivot_summary = pd.pivot_table(
    merged_df,
    values='Marks',
    index='Department',
    aggfunc=['sum', 'mean', 'count']
)

print("\n6. Pivot Table Summary:")
print(pivot_summary)
```

```python
import pandas as pd

# Create a DataFrame
data = {
    'Student_ID': [101, 102, 103, 104, 105],
    'Name': ['Rahul', 'Amit', 'Priya', 'Neha', 'Rohit'],
    'Age': [20, 21, 19, 22, 20],
    'Marks': [85, 78, 92, 88, 75],
    'Attendance': [90, 85, 95, 92, 80]
}

df = pd.DataFrame(data)

# Display the original DataFrame
print("Original DataFrame:")
print(df)

# Process the DataFrame by sorting marks in descending order
df = df.sort_values(by='Marks', ascending=False)

# Reset the index
df = df.reset_index(drop=True)

# Display the processed DataFrame
print("\nProcessed DataFrame:")
print(df)

# Export the DataFrame to a CSV file
df.to_csv("processed_students.csv", index=False)

print("\nData exported successfully to processed_students.csv")

# Verify the exported file by reading it
df_exported = pd.read_csv("processed_students.csv")

# Display the exported data
print("\nVerified Exported Data:")
print(df_exported)

# Verify the shape of the exported DataFrame
print("\nExported Data Shape:", df_exported.shape)

# Check whether the exported data matches the processed data
if df.equals(df_exported):
    print("Verification Successful: Exported data matches the processed data.")
else:
    print("Verification Failed: Data does not match.")
```

```python
# ==========================================================
# TASK 10: MINI E-COMMERCE DATA ANALYSIS PROJECT
# Libraries: NumPy, Pandas, Matplotlib, Seaborn
# Dataset: ecommerce.csv
# ==========================================================

# 1. Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 2. Load the Dataset
df = pd.read_csv("ecommerce.csv")

print("=" * 60)
print("1. ORIGINAL DATASET")
print("=" * 60)

print(df.head())
print("\nDataset Shape:", df.shape)
print("\nColumn Names:")
print(df.columns.tolist())


# 3. Data Inspection
print("\n" + "=" * 60)
print("2. DATA INSPECTION")
print("=" * 60)

print("\nDataset Information:")
df.info()

print("\nStatistical Summary:")
print(df.describe())

print("\nData Types:")
print(df.dtypes)

print("\nDuplicate Rows:", df.duplicated().sum())


# 4. Identify Missing Values
print("\n" + "=" * 60)
print("3. MISSING VALUES")
print("=" * 60)

print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())


# 5. Handle Missing Values
# Delivery_Date may be blank for undelivered or cancelled orders.
# Preserve these blanks because they have business meaning.

df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce")
df["Delivery_Date"] = pd.to_datetime(df["Delivery_Date"], errors="coerce")

# Remove duplicate records if present
df = df.drop_duplicates()

# Fill missing categorical values if any exist
for column in ["Product", "Category", "Payment_Method",
               "Order_Status", "City", "State", "Sales_Channel"]:
    df[column] = df[column].fillna("Unknown")

# Fill missing numerical values using the median
numeric_columns = [
    "Quantity",
    "Unit_Price_INR",
    "Discount_Percent",
    "Discount_Amount_INR",
    "Total_Amount_INR"
]

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

print("\n" + "=" * 60)
print("4. CLEANED DATASET")
print("=" * 60)

print(df.head())
print("\nRemaining Missing Values:")
print(df.isnull().sum())


# 6. NumPy Mathematical and Statistical Analysis
print("\n" + "=" * 60)
print("5. NUMPY STATISTICAL ANALYSIS")
print("=" * 60)

sales = df["Total_Amount_INR"].to_numpy()
quantity = df["Quantity"].to_numpy()

print("Total Sales (INR):", round(np.sum(sales), 2))
print("Average Order Value (INR):", round(np.mean(sales), 2))
print("Median Order Value (INR):", round(np.median(sales), 2))
print("Minimum Order Value (INR):", round(np.min(sales), 2))
print("Maximum Order Value (INR):", round(np.max(sales), 2))
print("Sales Standard Deviation:", round(np.std(sales), 2))
print("Total Quantity Sold:", int(np.sum(quantity)))


# 7. Business KPIs
print("\n" + "=" * 60)
print("6. BUSINESS KPIs")
print("=" * 60)

total_orders = df["Order_ID"].nunique()
total_customers = df["Customer_ID"].nunique()
total_products = df["Product"].nunique()

delivered_orders = (df["Order_Status"] == "Delivered").sum()
cancelled_orders = (df["Order_Status"] == "Cancelled").sum()
returned_orders = (df["Order_Status"] == "Returned").sum()

print("Total Orders:", total_orders)
print("Unique Customers:", total_customers)
print("Unique Products:", total_products)
print("Delivered Orders:", delivered_orders)
print("Cancelled Orders:", cancelled_orders)
print("Returned Orders:", returned_orders)

print(
    "Cancellation Rate (%):",
    round(cancelled_orders / total_orders * 100, 2)
)

print(
    "Return Rate (%):",
    round(returned_orders / total_orders * 100, 2)
)


# 8. Select Specific Columns
print("\n" + "=" * 60)
print("7. SELECTED COLUMNS")
print("=" * 60)

print(
    df[[
        "Order_ID",
        "Product",
        "Category",
        "Quantity",
        "Total_Amount_INR"
    ]].head(10)
)


# 9. Filter Records
print("\n" + "=" * 60)
print("8. FILTERING DATA")
print("=" * 60)

# Orders with value greater than INR 2000
high_value_orders = df[df["Total_Amount_INR"] > 2000]

print("\nOrders Above INR 2000:")
print(high_value_orders.head(10))

# Electronics orders
electronics_orders = df[df["Category"] == "Electronics"]

print("\nElectronics Orders:")
print(electronics_orders.head(10))

# Multiple conditions
high_value_delivered = df[
    (df["Total_Amount_INR"] > 2000) &
    (df["Order_Status"] == "Delivered")
]

print("\nHigh-Value Delivered Orders:")
print(high_value_delivered.head(10))


# 10. Sort Data
print("\n" + "=" * 60)
print("9. SORTING DATA")
print("=" * 60)

print("\nTop 10 Orders by Value:")
print(
    df.sort_values(
        by="Total_Amount_INR",
        ascending=False
    )[["Order_ID", "Product", "Total_Amount_INR"]].head(10)
)

print("\nLowest 10 Orders by Value:")
print(
    df.sort_values(
        by="Total_Amount_INR",
        ascending=True
    )[["Order_ID", "Product", "Total_Amount_INR"]].head(10)
)


# 11. GroupBy Analysis
print("\n" + "=" * 60)
print("10. CATEGORY-WISE ANALYSIS")
print("=" * 60)

category_analysis = df.groupby("Category").agg(
    Total_Orders=("Order_ID", "nunique"),
    Total_Quantity=("Quantity", "sum"),
    Total_Sales=("Total_Amount_INR", "sum"),
    Average_Order_Value=("Total_Amount_INR", "mean")
).sort_values(by="Total_Sales", ascending=False)

print(category_analysis.round(2))


# 12. Monthly Sales Analysis
print("\n" + "=" * 60)
print("11. MONTHLY SALES ANALYSIS")
print("=" * 60)

df["Month"] = df["Order_Date"].dt.to_period("M").astype(str)

monthly_sales = df.groupby("Month").agg(
    Total_Sales=("Total_Amount_INR", "sum"),
    Total_Orders=("Order_ID", "nunique")
)

print(monthly_sales.round(2))


# 13. Payment Method Analysis
print("\n" + "=" * 60)
print("12. PAYMENT METHOD ANALYSIS")
print("=" * 60)

payment_analysis = df.groupby("Payment_Method").agg(
    Total_Orders=("Order_ID", "nunique"),
    Total_Sales=("Total_Amount_INR", "sum")
).sort_values(by="Total_Sales", ascending=False)

print(payment_analysis.round(2))


# 14. City-Wise Sales Analysis
print("\n" + "=" * 60)
print("13. CITY-WISE SALES ANALYSIS")
print("=" * 60)

city_analysis = df.groupby("City").agg(
    Total_Orders=("Order_ID", "nunique"),
    Total_Sales=("Total_Amount_INR", "sum")
).sort_values(by="Total_Sales", ascending=False)

print(city_analysis.round(2))


# 15. Create a Pivot Table
print("\n" + "=" * 60)
print("14. PIVOT TABLE")
print("=" * 60)

pivot_table = pd.pivot_table(
    df,
    values="Total_Amount_INR",
    index="Category",
    columns="Order_Status",
    aggfunc="sum",
    fill_value=0
)

print(pivot_table.round(2))


# 16. Find Useful Insights
print("\n" + "=" * 60)
print("15. BUSINESS INSIGHTS")
print("=" * 60)

top_category = category_analysis["Total_Sales"].idxmax()
top_city = city_analysis["Total_Sales"].idxmax()
top_payment = payment_analysis["Total_Sales"].idxmax()

top_product = (
    df.groupby("Product")["Total_Amount_INR"]
    .sum()
    .idxmax()
)

highest_month = monthly_sales["Total_Sales"].idxmax()

print("Highest Revenue Category:", top_category)
print("Highest Revenue City:", top_city)
print("Most Valuable Payment Method:", top_payment)
print("Highest Revenue Product:", top_product)
print("Highest Sales Month:", highest_month)


# 17. Data Visualization
sns.set_theme(style="whitegrid")

# Category-wise sales chart
plt.figure(figsize=(10, 6))

category_analysis["Total_Sales"].sort_values().plot(
    kind="barh"
)

plt.title("Category-Wise Sales")
plt.xlabel("Total Sales (INR)")
plt.ylabel("Category")
plt.tight_layout()
plt.savefig("category_wise_sales.png", dpi=300)
plt.show()


# Monthly sales chart
plt.figure(figsize=(12, 6))

monthly_sales["Total_Sales"].plot(
    kind="line",
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales (INR)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("monthly_sales_trend.png", dpi=300)
plt.show()


# Order status chart
plt.figure(figsize=(8, 5))

df["Order_Status"].value_counts().plot(
    kind="bar"
)

plt.title("Order Status Distribution")
plt.xlabel("Order Status")
plt.ylabel("Number of Orders")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig("order_status_distribution.png", dpi=300)
plt.show()


# 18. Export the Processed Dataset
print("\n" + "=" * 60)
print("16. EXPORT PROCESSED DATA")
print("=" * 60)

df.to_csv("cleaned_ecommerce_dataset.csv", index=False)

print("Cleaned dataset exported successfully.")


# 19. Verify the Exported File
verified_df = pd.read_csv("cleaned_ecommerce_dataset.csv")

print("\nExported Dataset Shape:", verified_df.shape)
print("\nExported Dataset Preview:")
print(verified_df.head())

print("\nVerification Result:")
print("Export successful:", len(verified_df) == len(df))
print(
    "Row count matches:",
    len(verified_df) == len(df)
)

print("\nProject Completed Successfully.")
```
